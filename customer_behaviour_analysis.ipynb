```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#create a column age_group
labels= ['young Adult','Adult','Middle-aged','Senior']
df['age_group'] =pd.qcut(df['age'],q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create a purchase_freduency_day

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90,
        
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus
import pandas as pd

# Step 1 — connect
conn_str = quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=Rakesh\\SQLEXPRESS;"
    "DATABASE=customer_behaviour;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={conn_str}")

# Step 2 — write your DataFrame to SQL
df.to_sql(
    name      = "customer",    # table name in SQL Server
    con       = engine,
    if_exists = "replace",     # replace if table exists
    index     = False          # don't write row numbers
)

print(f"Done! {len(df)} rows written to SQL Server.")
```

```python
# Read it back to verify
result = pd.read_sql("SELECT * FROM customer", engine)
print(f"Rows in SQL: {len(result)}")
result.head()
```
